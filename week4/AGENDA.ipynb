## Deploying and Monitoring Agents on Databricks

This course covers the end-to-end lifecycle for deploying and monitoring generative AI agents on Databricks. Participants will learn how to deploy agents as Databricks Apps using Declarative Automation Bundles (DABs), integrate tools via the Model Context Protocol (MCP), instrument agents with MLflow Tracing, and evaluate production quality using scorers, multi-turn judges, and online evaluation. Through hands-on demos and labs, participants will gain practical experience building, observing, and monitoring production-grade AI agents on the Databricks platform.

----

## Prerequisites
The content was developed for participants with these skills/knowledge/abilities:
- Familiarity with Databricks workspace and notebooks
- Familiarity with Unity Catalog
- Experience building agents using the OpenAI Agents SDK
- Basic knowledge of MLflow and Python
- Familiarity with GenAI agent concepts (LLM calls, tool invocation, retrieval)

## Course Agenda
The following modules are part of the **Generative AI Application Deployment and Monitoring** course by **Databricks Academy**.

| # | Type | Notebook | Description |
|---|------|----------|-------------|
| 1 | Lecture | [Agent Deployment on Databricks]($./1_Lecture_Agent_Deployment_on_Databricks) | Agent lifecycle, Databricks Apps, DABs, deployment patterns, resource injection |
| 2 | Demo | [Deploying an Observable Agent]($./2_Demo_Deploying_an_Observable_Agent) | Progressively build an agent: deploy as an App, add MLflow tracing, integrate MCP tools |
| 3 | Lecture | [Tool Integration and Observability]($./3_Lecture_Tool_Integration_and_Observability) | Model Context Protocol, MCP server types, MCP lifecycle, MLflow Tracing foundations |
| 4 | Demo | [Tracing for Production Agents]($./4_Demo_Tracing_for_Production_Agents) | Manual spans, custom attributes, multi-turn session grouping, programmatic trace search |
| 5 | Lecture | [Production Evaluation and Monitoring]($./5_Lecture_Production_Evaluation_and_Monitoring) | Scorer types, custom scorers, multi-turn judges, sampling strategies, scorer lifecycle |
| 6 | Lab | [Online Evaluation with Multi-Turn Conversations]($./6_Lab_Online_Evaluation_with_Multi-Turn_Conversations) | Deploy judges, create custom scorers, tune sampling rates, run the metrics computation job |
| 7 | Lab | [Backfilling and Querying Multi-Turn Conversations]($./7_Lab_Backfilling_and_Querying_Multi-Turn_Conversations) | Backfill historical traces, query traces and assessments in Unity Catalog with SQL, scorer cleanup |
| 8 | Lecture | [Appendix]($./8_Lecture_Appendix:_Agent_Bricks_CLI) | Agent Bricks CLI (Beta): init, dev, and deploy lifecycle, agent.toml, managed tools, memory, and sessions |
