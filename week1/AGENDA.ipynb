## Building RAG Agents with Agent Bricks

This course provides hands-on training for building retrieval agents on Databricks. Participants will learn to create Genie Agents for unstructured data, perform information retrieval with Lakebase Search, parse and transform unstructured documents into structured data using AI Functions, chunk text for semantic retrieval, and build AI Search indexes with multiple search modes.

---

## Prerequisites

- Basic SQL knowledge for querying tables and using built-in functions
- Basic Python programming experience
- Familiarity with the Databricks workspace UI (e.g. navigating Unity Catalog)
- Understanding of Unity Catalog concepts including catalogs, schemas, and volumes
- Familiarity with fundamental LLM concepts (e.g. embeddings, retrieval-augmented generation)
- Understanding of structured vs unstructured data

### Technical Requirements

Please review the following requirements before starting the course:

- Use **Serverless Compute (environment version 6)** for running all demo and lab notebooks
- Access to **`ai_parse_document()`**, **`ai_prep_search()`**, **`ai_classify()`**, and **`ai_extract()`** functions (Beta features)
- Access to **Databricks Agent Bricks** (Genie Agents)
- A pre-provisioned **Lakebase Postgres** project and branch (provisioned during workspace setup)
- A pre-created **AI Search endpoint** (`vs_all_demo`, provisioned during workspace setup)
- Access to **Foundation Model APIs** for managed embeddings (`databricks-gte-large-en`)
