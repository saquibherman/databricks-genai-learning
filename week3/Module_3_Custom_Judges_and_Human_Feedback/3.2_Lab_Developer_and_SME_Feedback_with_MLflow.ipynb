### C5. Viewing SME Feedback in a Notebook

The MLflow trace is available using the UI, but we can also view the assessment from the SME programmatically. The code below retrieves your most recent traces from the app's experiment using `search_traces()`, filtered to your user. After running the next cell, scroll and expand the **assessments** column for the record returned.

```python
import pandas

mlflow.set_tracking_uri("databricks")

## Resolve the app's experiment ID
app_exp = mlflow.get_experiment_by_name(app_experiment_path)
print("App experiment ID:", app_exp.experiment_id)

## Query your traces from the app, newest first
traces_df = mlflow.search_traces(
    locations=[str(app_exp.experiment_id)],          ## must be str
    filter_string=f"metadata.`mlflow.trace.user` = '{username}'",
    order_by=["timestamp_ms DESC"],
    max_results=5,
    return_type="pandas",
)
traces_df.head(1)
```
